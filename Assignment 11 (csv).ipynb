```python
import csv
```

```python
open_file = open ("student data - student data.csv", mode = 'r')
open_file
```

```python
reader = csv.reader(open_file)
reader
```

```python
data = list(reader)
data
```

```python
data[0]
```

```python
data[0].append('Grade')  

for i in range(1, len(data)):
    percentage = float(data[i][5])
    if percentage >= 80:
        grade = 'A'
    elif percentage >= 70:
        grade = 'B'
    elif percentage >= 60:
        grade = 'C'
    elif percentage >= 50:
        grade = 'D'
    else:
        grade = 'F'
    data[i].append(grade)
```

```python
data
```

```python
updated = open('students grade.csv', mode='w', newline='') 
writer = csv.writer(updated)
writer.writerows(data)
updated.close()
```
